```python
import matplotlib.pyplot as plt
```

```python
x= [1,2,3,4]
y= [2,4,6,8]
plt.plot(x,y)
plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

```python
#Bar Chart
months = ['Jan','Feb','Mar','April']
sales = [250,300,400,350]
plt.bar(months,sales,color = 'skyblue')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

```python
plt.barh(months,sales,color = 'orange')
plt.title('Monthly sales (Horizontal)')
plt.xlabel('Sales')
plt.ylabel('Months')
plt.show()
```

```python
#Scatter Plot (Comparing Height vs Weight)
height = [150 , 160 , 170 , 180]
weight = [50 , 60 , 65 , 75]
plt.scatter(height,weight,color = 'green')
plt.title('Height vs Weight')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
#Pie Chart(Market Share)
brands = ['Brand A' , 'Brand B' , 'Brand C']
shares = [45 , 30 , 25]
plt.pie(shares , labels=brands , autopct = '%1.1f%%' , startangle = 90)
plt.title('Market Shares')
plt.axis('equal')
plt.show()
```

```python
#Multiple Plot in one figure
x = [1 , 2 , 3 , 4]
y1 = [1 , 4 , 9 , 16]
y2 = [1 , 2 , 3 , 4]
plt.plot(x , y1 , label = 'Square')
plt.plot(x , y2 , label = 'Linear')
plt.legend()
plt.title('Multiple Lines')
plt.show()
```

```python
#Subplots
fig,axs = plt.subplots(2,2)

axs[0,0].plot(x,y1)
axs[0,0].set_title('Square')

axs[0,1].bar(months,sales)
axs[0,1].set_title('Sales')

axs[1,0].scatter(height,weight)
axs[1,0].set_title('Scatter')

axs[1,1].pie(shares , labels=brands , autopct = '%1.1f%%' , startangle = 90)
axs[1,1].set_title('Pie')

```
