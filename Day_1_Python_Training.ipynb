```python
class Student:
    def __init__(self, name, roll_number, marks):
        self.name = name
        self.roll_number = roll_number

        self.marks = marks
        self.total = sum(marks)
        self.percentage = (self.total / (5 * 100)) * 100

    def check_admission(self):
        print(f"\n--- Student Result Summary ---")
        print(f"Name: {self.name}")
        print(f"Roll Number: {self.roll_number}")
        print(f"Total Marks: {self.total} / 500")
        print(f"Percentage: {self.percentage:.2f}%")

        if self.percentage >= 85:
            print("Status: Admission Approved!")
        else:
            print("Status: Admission Rejected (Percentage below 85%)")

# Input gathering
name = input("Enter Student Name: ")
roll_no = input("Enter Roll Number: ")
marks = []
for i in range(1, 6):
    mark = float(input(f"Enter marks for Paper {i} (out of 100): "))
    marks.append(mark)

# Create Student object and evaluate admission status
student_obj = Student(name, roll_no, marks)
student_obj.check_admission()
```
